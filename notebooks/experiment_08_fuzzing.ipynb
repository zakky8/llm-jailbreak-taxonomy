```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print('Experiment 08: Fuzzing-Based Attack Taxonomy initialized')

data = {
    'Pattern': ['FZ-01 Synonym Mutation', 'FZ-02 Semantic Transform', 'FZ-03 Crossover Mutation'],
    'ASR': [99, 97, 94],
    'Reference': ['JBFuzz 2025', 'JBFuzz 2025', 'JBFuzz 2025']
}
df = pd.DataFrame(data)
print(df.to_string())
```

```python
models = ['GPT-4o', 'Gemini 2.0', 'DeepSeek-V3']
baseline = [45, 42, 50]
jbfuzz = [99, 98, 97]

x = np.arange(len(models))
width = 0.35
fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(x - width/2, baseline, width, label='Baseline Human Attack', color='#3498db')
ax.bar(x + width/2, jbfuzz, width, label='JBFuzz Automated', color='#e74c3c')
ax.set_ylabel('Attack Success Rate (%)')
ax.set_title('JBFuzz vs Baseline ASR by Model (JBFuzz, 2025)')
ax.set_xticks(x)
ax.set_xticklabels(models)
ax.legend()
ax.set_ylim(0, 115)
plt.tight_layout()
plt.show()
```
