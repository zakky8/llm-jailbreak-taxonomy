## 1. Fuzzing Mutation Strategies

Automated fuzzing-based attacks (e.g., JBFuzz) typically apply three types of mutations to an initial seed prompt:
- **Semantic Mutation:** Replacing keywords with low-frequency synonyms.
- **Format Mutation:** Wrapping payloads in obscure markdown or JSON structures.
- **Crossover Mutation:** Combining two successful bypasses into a hybrid prompt.

```python
@dataclass
class FuzzPattern:
    id: str
    mutation_type: str
    mechanism: str
    detectability: str
    priority: str

fuzz_patterns = [
    FuzzPattern('FZ-01', 'semantic', 'synonym_substitution', 'low', 'CRITICAL'),
    FuzzPattern('FZ-02', 'structural', 'format_obfuscation', 'medium', 'HIGH'),
    FuzzPattern('FZ-03', 'hybrid', 'crossover_evolution', 'very_low', 'CRITICAL')
]

df = pd.DataFrame([asdict(p) for p in fuzz_patterns])
df
```
