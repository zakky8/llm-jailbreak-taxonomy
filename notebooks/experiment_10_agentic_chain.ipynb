## 1. Agentic Attack Mechanisms

- **Memory Poisoning:** Inserting adversarial context into long-term memory (RAG/Vector DB) that triggers a jailbreak when the model retrieves it later.
- **Tool Argument Injection:** Exploiting models that call external tools (Python, Search, Terminal) by injecting control characters or malicious parameters into valid tool calls.
- **Session Persistence:** Designing attacks that survive model "resets" by utilizing the agent's environment state.
