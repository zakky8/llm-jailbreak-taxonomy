## Objective
Document multi-agent pipeline attacks including tool chain hijacking and cross-session memory poisoning.

## Patterns: AG-01, AG-02

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Experiment 10: Agentic Chain Exploitation Taxonomy initialized')

patterns = [
    {'id': 'AG-01', 'name': 'Tool Chain Hijack', 'persistence': 'session', 'assumption': 'tool_output_trust', 'severity': 'CRITICAL'},
    {'id': 'AG-02', 'name': 'Memory Poisoning', 'persistence': 'cross_session', 'assumption': 'memory_integrity', 'severity': 'CRITICAL'},
]
df = pd.DataFrame(patterns)
print(df.to_string())

fig, ax = plt.subplots(figsize=(8, 4))
attack_types = ['Single Turn\nChat', 'Multi Turn\nChat', 'Agentic\nSingle Tool', 'Agentic\nTool Chain', 'Memory\nPoisoning']
complexity = [1, 2, 3, 4, 5]
asr = [30, 55, 65, 80, 85]
ax.plot(complexity, asr, 'o-', color='#e74c3c', linewidth=2, markersize=8)
for i, (x, y, label) in enumerate(zip(complexity, asr, attack_types)):
    ax.annotate(label, (x, y), textcoords='offset points', xytext=(0, 12), ha='center', fontsize=8)
ax.set_xlabel('Attack Complexity')
ax.set_ylabel('Estimated ASR (%)')
ax.set_title('Attack Success Rate vs Complexity\n(Agentic attacks exploit increasing attack surface)')
ax.set_ylim(0, 110)
plt.tight_layout()
plt.show()
print('Key insight: AG-02 memory poisoning is the first cross-session persistence attack in this taxonomy')
```
