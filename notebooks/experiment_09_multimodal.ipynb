## Objective
Document cross-modal attack vectors exploiting gaps between text and image safety classifiers.

## Patterns: MM-01, MM-02

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Experiment 09: Multimodal Injection Taxonomy initialized')

patterns = [
    {'id': 'MM-01', 'name': 'Image-Text Mismatch', 'vector': 'cross_modal', 'assumption': 'Independent modality evaluation'},
    {'id': 'MM-02', 'name': 'Visual Prompt Injection', 'vector': 'image_embedded', 'assumption': 'Text-only instruction processing'},
]
df = pd.DataFrame(patterns)
print(df.to_string())

fig, ax = plt.subplots(figsize=(8, 4))
categories = ['Text-Only\nClassifier', 'Image-Only\nClassifier', 'Cross-Modal\nClassifier']
coverage = [85, 80, 45]
colors = ['#2ecc71', '#3498db', '#e74c3c']
ax.bar(categories, coverage, color=colors, width=0.5)
ax.set_ylabel('Safety Coverage (%)')
ax.set_title('Safety Classifier Coverage by Modality\n(Cross-modal attacks exploit the gap)')
ax.set_ylim(0, 110)
for i, v in enumerate(coverage):
    ax.text(i, v + 2, f'{v}%', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```
