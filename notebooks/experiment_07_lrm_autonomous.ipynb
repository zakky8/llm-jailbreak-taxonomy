## 1. Taxonomy Framework — LRM Mechanisms

Unlike human-engineered prompts, LRM-generated attacks exploit the internal reasoning capacity of the model. This creates an "Adversarial Loop" where the model self-corrects its prompts until the target model bypasses its filter.

**Mechanisms targeted:**
- **Reasoning Hijacking:** Forcing the target model's CoT into a logic state where refusal is logically inconsistent.
- **Semantic Variability:** Iterating synonym space at scale until a low-probability bypass is hit.
- **Structured Planning:** Breaking a jailbreak into multiple logical phases (Recon -> Probe -> Exploit).

```python
@dataclass
class LRMAttackPattern:
    id: str
    subcategory: str
    mechanism: str
    sophistication: str
    exploited_assumption: str
    asr_reported: str

lrm_patterns = [
    LRMAttackPattern(
        id='LRM-01',
        subcategory='autonomous_planning',
        mechanism='reasoning_model_exploitation',
        sophistication='advanced',
        exploited_assumption='human_speed_adversary',
        asr_reported='97%'
    ),
    LRMAttackPattern(
        id='LRM-02',
        subcategory='cot_hijacking',
        mechanism='logic_state_corruption',
        sophistication='advanced',
        exploited_assumption='linear_instruction_processing',
        asr_reported='92%'
    ),
    LRMAttackPattern(
        id='LRM-03',
        subcategory='self_refinement',
        mechanism='iterative_optimization',
        sophistication='advanced',
        exploited_assumption='static_prompt_defense',
        asr_reported='95%'
    )
]

df = pd.DataFrame([asdict(p) for p in lrm_patterns])
print(f'Documented {len(lrm_patterns)} LRM attack patterns')
df
```
