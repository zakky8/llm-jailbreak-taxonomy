## Objective
Document and analyze autonomous jailbreak attacks executed by large reasoning models (LRMs) such as DeepSeek-R1, Gemini 2.5 Flash, and Grok 3 Mini. Shah et al. (2025) demonstrated these models achieve 97% ASR across 9 target models.

## Exploited Alignment Assumption
Safety alignment assumes a human adversary with limited iteration speed. LRM attackers iterate at machine speed with systematic reasoning — invalidating this assumption entirely.

## Patterns Covered
- LRM-01: Autonomous planning and execution
- LRM-02: Chain-of-thought manipulation
- LRM-03: Self-refinement exploit

```python
import pandas as pd
import matplotlib.pyplot as plt
from dataclasses import dataclass, asdict
from typing import List, Optional

print('Research environment initialized')
print('Experiment 07: LRM Autonomous Attack Taxonomy')
```

```python
@dataclass
class LRMAttackPattern:
    pattern_id: str
    subcategory: str
    mechanism: str
    asr_reported: str
    reference: str
    notes: str

patterns = [
    LRMAttackPattern('LRM-01','autonomous_planning','reasoning_model_exploitation','97% across 9 models','Shah et al. (2025)','LRM autonomously plans multi-turn jailbreak strategy'),
    LRMAttackPattern('LRM-02','chain_of_thought_manipulation','cot_hijacking','Not yet quantified','Shah et al. (2025)','Adversarial manipulation of CoT reasoning steps'),
    LRMAttackPattern('LRM-03','self_refinement_exploit','iterative_bypass','Not yet quantified','Emerging 2026','Self-refinement loop iterates until bypass succeeds'),
]

df = pd.DataFrame([asdict(p) for p in patterns])
print(df[['pattern_id','subcategory','asr_reported']].to_string())
```

```python
categories = ['Human\nEngineered', 'LRM\nAutonomous']
asr_values = [45, 97]
colors = ['#3498db', '#e74c3c']

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(categories, asr_values, color=colors, width=0.4)
ax.set_ylabel('Attack Success Rate (%)')
ax.set_title('Human vs LRM Autonomous Attack Success Rates\n(Shah et al., 2025)')
ax.set_ylim(0, 115)
for bar, val in zip(bars, asr_values):
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 2, f'{val}%', ha='center', va='bottom', fontweight='bold', fontsize=14)
ax.axhline(y=50, color='gray', linestyle='--', alpha=0.5, label='50% threshold')
ax.legend()
plt.tight_layout()
plt.show()
print('LRM attacks achieve 2.16x higher ASR than human-engineered attacks')
```
